```python
# Download all stock prices
import pandas as pd
from vnstock import Vnstock

df = pd.read_excel("Stock_List.xlsx")
stock_data = {}

# Self-input the timeframe
start_date = input("Start date (YYYY-MM-DD): ").strip()
end_date = input("End date (YYYY-MM-DD): ").strip()

for symbol in df["Stock_code"]:
    symbol = str(symbol).strip().upper()
    try:
        stock = Vnstock().stock(symbol=symbol, source="VCI")
        price_data = stock.quote.history(
            start=start_date,
            end=end_date,
            interval="1D"
        )

        price_data["time"] = pd.to_datetime(price_data["time"])

        price_data["Stock_code"] = symbol

        stock_data[symbol] = price_data
        print(f"✅ Loaded data for {symbol}")
        
    except Exception as e:
        print(f"❌ Error with {symbol}: {e}")

# Export to csv
if stock_data:
    export_df = pd.concat(stock_data.values(), ignore_index=True)

    # 🔹 SORT: Stock_code first, then time
    export_df = export_df.sort_values(
        by=["Stock_code", "time"],
        ascending=[True, True]
    )

    export_df.to_csv(
        f"VN_Stock_Prices_{start_date}_{end_date}.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print("📁 Exported to CSV successfully")


```

```python
# Download income statement
from vnstock import Finance

# Load stock list
df = pd.read_excel("Stock_List.xlsx")

all_income_statements = []

for symbol in df["Stock_code"]:
    symbol = str(symbol).strip().upper()
    try:
        finance = Finance(symbol=symbol, source="VCI")
        income_df = finance.income_statement(period="year", lang="en")

        all_income_statements.append(income_df)

        print(f"✅ Retrieved income statement for {symbol}")

    except Exception as e:
        print(f"❌ Error with {symbol}: {e}")

# Export into csv
if all_income_statements:
    final_df = pd.concat(all_income_statements, ignore_index=True)

    final_df.to_csv(
        "VN_Income_Statements_All.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print("📁 Exported all income statements to CSV")
```
